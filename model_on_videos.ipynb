# **Generalized function:**
same pipeline- 5 videos 3 models: 11n 11m and our pre trained

```python
def run_full_tracking_pipeline(model, video_path, gt_json_path, pred_json_output_json, output_video_path=None):
    import os

    # Derive default video output path if not provided
    if output_video_path is None:
        output_video_path = f"tracked_{os.path.basename(video_path)}"

    # Derive predicted .txt output from predicted .json
    pred_txt_output = pred_json_output_json.replace(".json", ".txt")

    # Derive GT .txt output
    gt_txt_output = gt_json_path.replace(".json", ".txt")

    print("🔍 Running detection + tracking...")
    run_ocsort_save_json(model, video_path, output_video_path, pred_json_output_json)

    print("📄 Converting ground-truth annotations to MOT format...")
    convert_json_to_mot(gt_json_path, gt_txt_output)

    print("📄 Converting predictions to MOT format...")
    convert_json_to_mot(pred_json_output_json, pred_txt_output)

    print("📊 Computing tracking metrics...")
    compute_tracking_metrics(gt_txt_output, pred_txt_output)

    print("📈 Calculating mAP (COCO style)...")
    map_5095, map_50 = map_calc_coco(
        gt_json_path=gt_json_path,
        pred_json_path=pred_json_output_json,
        video_path=video_path
    )
    print(f"mAP@0.50: {map_50:.4f}")
    print(f"mAP@0.50:0.95: {map_5095:.4f}")

    print("✅ Done!")
```

```python
run_full_tracking_pipeline(model, "/content/videos /birds_beach_856799-hd_1920_1080_25fps.mp4", "/content/ground truth/gt_birds_beach.json", "birds_beach_results.json")
```
